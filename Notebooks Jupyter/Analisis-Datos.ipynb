```python
import pandas as pd
import geopandas as gpd
import matplotlib.pyplot as plt

SHAPEFILE = "../dashboard_alerts-shapefile/dashboard_alerts_shapefile.shp"

gdf = gpd.read_file(SHAPEFILE)

print("\n=== Informacion Basica ===")
#print(gdf.head())
print("\nColumnas:")
print(gdf.columns.tolist())
print("\nForma de la tabla:")
print(gdf.shape)
print("\nCRS:")
print(gdf.crs)
print("\nInformacion basica y tipos :")
print(gdf.info())
```

```python
print("=" * 120)
print(f"Registros: {len(gdf)}")
print(f"Variables: {len(gdf.columns)}")
print("=" * 120)

for col in gdf.columns:

    print(f"\n{'='*120}")
    print(f"Variable: {col}")
    print(f"{'='*120}")

    # Tipo de dato
    print(f"Tipo: {gdf[col].dtype}")

    # Nulos
    nulos = gdf[col].isna().sum()
    nulos += gdf[col].isnull().sum()
    porcentaje_nulos = (nulos / len(gdf)) * 100

    print(f"Nulos: {nulos} ({porcentaje_nulos:.2f}%)")

    # Valores únicos
    uniques = gdf[col].nunique(dropna=True)

    print(f"Valores únicos: {uniques}")

    # Conteo de cada valor
    print("\nFrecuencia de valores:")

    value_counts = gdf[col].value_counts(dropna=False)

    print(value_counts)

    # Estadísticas para variables numéricas
    if pd.api.types.is_numeric_dtype(gdf[col]):

        print("\nEstadísticas:")

        print(gdf[col].describe())

    else:

        print("\nPrimeros valores únicos:")

        print(gdf[col].dropna().unique()[:20])
```
